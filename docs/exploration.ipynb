## Generating Virtualizarr representation of model output to then make model catalog

```python


out = Path("/Users/kthyng/cstar-forge-data/cstar-blueprint-run/cson_roms-marbl_v0.1_GOM-Offline_noPIO_6procs/output_dt450")
GRID = Path("/Users/kthyng/cstar-forge-data/cstar-blueprint-run/cson_roms-marbl_v0.1_GOM-Offline_noPIO_6procs/input/input_datasets/cson_roms-marbl_v0_1_GOM-Offline_noPIO_6procs_grid.nc")
make_kerchunk(sorted(out.glob("output_bgc.*.nc")), "refs/gom_bgc.parquet",
                grid=GRID)

cat = new_catalog(title="GOM offline run")
add_source(cat, "GOM_bgc", "refs/gom_bgc.parquet")
save(cat, "catalogs/gom.yaml")
```

## Catalog for Ocean SODA

https://www.ncei.noaa.gov/data/oceans/ncei/ocads/metadata/0220059.html

```python
loc = "https://www.ncei.noaa.gov/data/oceans/ncei/ocads/data/0220059/OceanSODA_ETHZ-v2025.OCADS.01-1982-2024.nc"

cat = new_catalog(title="OceanSODA")
add_source(cat, "OceanSODA", f"simplecache://::{loc}", storage_options={"simplecache": {"same_names": True}})
save(cat, "catalogs/oceansoda.yaml")
```
