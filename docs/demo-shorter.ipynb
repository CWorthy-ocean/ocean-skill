## Catalog management

You need to make a catalog to represent your model. Examples are available in catalogs.ipynb and I might have already made your catalog for you. The goal is to make this step very easy.

You hopefully don't need to make your own data catalog but the goal is for that to be pretty easy too, once I've made some from a lot of different sources. If the dataset is not easy to load then it will be harder to make the catalog.

Some catalog options are here:
* `osk.find()` with options of "text", "variable", "name", "featureTyoe", "catalog", "climatology", "bbox", "time", "resolution", "cadence", "vertical"
* `osk.describe()` with either a catalog name or data source name as a string to see the metadata
* `osk.read()` with source name reads in the source to an xarray Dataset or pandas DataFrame.

## Compare Two Models

### Several Variables at One Time

```python
multi_vars = osk.compare(
    reference="dev",
    test="dev_marblsub",
    variables=["nitrate", "phosphate", "silicate", "oxygen"],
    depths=["surface"],
    select={"T": '2010-02-19T00:00:00'},
)
fig = multi_vars.plot(title=f"dev vs. dev_marblsub (Surface, 2010-02-19T00:00:00)", zoom=1.5)
```

### Calculate metrics between models

```python
metrics = osk.compare(
    reference="dev",
    test="dev_marblsub",
    variables=["silicate"],
    depths=["surface"],
    # select={"T": '2010-02-19T00:00:00'},
    # aggregate={"time": {"resample": "1MS", "reduce": "mean"}},
    over="time",
)
fig = metrics.plot(title=f"dev vs. dev_marblsub (Surface Silicate)", zoom=1.5)
```

### One Variable Monthly Means

```python
# NEW VERSION
monthly_means = osk.compare(
    reference="dev", test="dev_marblsub",
    variables=["silicate"], depths=["surface"],
    times={"resample": "1MS", "reduce": "mean"},
)
fig = monthly_means.plot(title="dev vs. dev_marblsub (Surface, Monthly Means)", zoom=1.5)
# physics.movie(save="silicate_months.mp4")
```

### Time Series

```python
import ocean_skill as osk

point = osk.compare(
    reference="dev", test="dev_marblsub", variables=["temperature","salinity"],
    select={"lon": -144.25, "lat": 50.0},#, "time": slice("2012-01", "2012-12")},
)
fig = point.plot()                     # reference solid, test dashed
# point.plot(renderer="holoviews")
```
